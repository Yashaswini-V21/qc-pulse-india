```python
import os; print(os.listdir('../data/raw/'))
```

```python
df_blinkit = pd.read_csv('../data/raw/BlinkIT Grocery Data.csv', 
                          encoding='latin-1', 
                          on_bad_lines='skip',
                          sep=',',
                          engine='python')
print('Blinkit ✅', df_blinkit.shape)
```

```python
# Test 2
df_zepto = pd.read_csv('../data/raw/zepto_v1.csv', encoding='latin-1', on_bad_lines='skip', engine='python')
print('Zepto ✅', df_zepto.shape)
```

```python
# Test 3
df_groceries = pd.read_csv('../data/raw/Groceries_dataset.csv', encoding='latin-1', on_bad_lines='skip', engine='python')
print('Groceries ✅', df_groceries.shape)
```

```python
# Test 4
df_bigbasket = pd.read_csv('../data/raw/BigBasket Products.csv', encoding='latin-1', on_bad_lines='skip', engine='python')
print('BigBasket ✅', df_bigbasket.shape)
```

```python
# Cell 3 - See all columns
for name, df in [('Blinkit', df_blinkit), ('Zepto', df_zepto), 
                  ('Groceries', df_groceries), ('BigBasket', df_bigbasket)]:
    print(f'\n{"="*50}')
    print(f'📦 {name} | {df.shape[0]:,} rows × {df.shape[1]} cols')
    print(f'Columns: {df.columns.tolist()}')
    display(df.head(3))
```

```python
# Cell 4 - Null check
for name, df in [('Blinkit', df_blinkit), ('Zepto', df_zepto),
                  ('Groceries', df_groceries), ('BigBasket', df_bigbasket)]:
    print(f'\n{"="*50}')
    print(f'🩺 {name} — Nulls:')
    nulls = df.isnull().sum()
    print(nulls[nulls > 0] if nulls.sum() > 0 else 'No nulls ✅')
```
